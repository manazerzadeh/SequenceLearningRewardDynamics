```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
x1 = np.random.rand(1000) + 50
x2 = np.random.rand(1000) + 2

y = 0.5 * x1 + 0.5 * x2 + np.random.normal(0, 0.01, 1000) + 10
```

```python
# plot the data cloud
sns.scatterplot(x=x1, y=x2, hue=y, palette='viridis')

```

```python
# fit using lstsq with intercept
A = np.vstack([x1, x2, np.ones(len(x1))]).T
coef, _, _, _ = np.linalg.lstsq(A, y, rcond=None)
print(f'Coefficients: x1: {coef[0]}, x2: {coef[1]}, intercept: {coef[2]}')
```

```python
# regress out the readout component
y_pred = A @ coef
residuals = y - y_pred
sns.scatterplot(x=x1, y=x2, hue=residuals, palette='viridis')
```

```python
A.shape
```

```python
# regress out the readout component
y_pred = A[:, :-1] @ coef[:-1]
residuals = y - y_pred
sns.scatterplot(x=x1, y=x2, hue=residuals, palette='viridis')
```
